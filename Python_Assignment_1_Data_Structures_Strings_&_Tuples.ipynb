```python
# String Slicing and Indexing
```

```python
s = "Hello Zara", "Welcome to Python programming"
```

```python
# a. Print the first character of the string
print("a.", s[0])
```

Python indexing starts at 0,so s[0] is the first item.
a.H means s[0] returned H,the first character of "Hello Zara".

```python
# b. Print the last character of the string
print("b.", s[-1])
```

Negative indices count from the end.s[_1] is the last item.
b.g means s[_1] returnes g,the last letter of "Welcome to Python programming"

```python
# c. print the fiirst 5 characters of the string
print("c.", s[:5])
```

Slicing uses the format s[start:stop].The start is included and the stop is excluded.When the start is left blank,Python begins at index 0,so s[:5] is the same as s[0:5].That takes indicates 0,1,2,3,4,which
are H,e,l,l,o.

```python
# d. Print the last 11 characters of the string
print("d.", s[-11:])
```

s[_11:] starts at the 11th character from the end and continues to the end of the string.Counting back from the end of "Welcome to Python programming".

```python
# e. Print the string in reverse
print("e.", s[::-1])
```

The full slice format is s[start:stop:step].Here start and stop are both blank,so Python covers the whole string.Python starts at the last character and works towars the first.

```python
# f.Use slicing and print the word "Python" from the existing string
print("f.", s[-18:-12])
```

s[_18:12] starts at the 18th character from the end and stops at the 12th from the end.

```python
# String Methods
```

```python
strM = "Python beginner tutorial"
# a. Sentence to uppercase
print("a.", strM.upper())
```

strM = "Python beginner tutorial" stores the sentence in a
variable named strM.strM.upper() is a string method that returns
a new copy of the string with every letter converted to
uppercase.

```python
# b. Sentence to lowercase
print("b.", strM.lower())
```

strM.lower() is a string method that returns a new copy of
strM with every letter converted to lowercase.

```python
# c. Use capitalize and return the sentence to the original input form
print("c.", strM.lower().capitalize())
```

capitalize()then makes the first character uppercase and forces the rest tolowercase,giving "Python beginner tutorial".

```python
# d. Count the number of occurences of character 't' in the string
print("d.", strM.count('t'))
```

strM.count('t')scans the string and returns howmany times the
substring 't' appears.In "Python beginner tutorial",
the lowercase t appears in:
Python(1),tutorial at the start(2),tutorial in the middle(3).

```python
# e. Replace all occurences of "Python" with "Machine Learning" in the input string strM= "Python beginner tutorial"
print("e." ,strM.replace ("Python", "Machine Learning"))
```

strM.replace ("Python","Machine Learning") searches the string for
every occurences of "Python" and returns a new string with each one strapped for "Machine Learning".

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate
t_combine = t1 + t2
print("a.", t_combine)
```

t1 and t2 are two separate tuples.The + operator joining two tuples.
t1+t2 concatenates them,creating a new tuple.

```python
# b. Repeat 3 times
print("b.", t_combine * 3)
```

*3 builds a new tuple by chaining three copies of it end to end.
* operator means repetation.

```python
# c. 3rd element (index 2)
print("c.", t_combine[2])
```

Indexing starts at 0,so index 2 is the 3rd element.

```python
# d. First three elements
print("d.", t_combine[:3])
```

Slicing is[start:stop],where stop is excluded.
Start means from the beginning,so [:3] takes indicates 0,1,2.

```python
# e. Last three elements
print("e.", t_combine[-3:])
```

Negative indicates count from the end:
_1 is the last element,_3 is the third from last.
stop means to the end,so [_3:] takes the last three elements.
